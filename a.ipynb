```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import requests
import time
import firebase_admin
from firebase_admin import credentials, firestore
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime, timezone, timedelta  # ✅ Using datetime without pytz

# Initialize Firebase
cred = credentials.Certificate("credentials/firebase-service-account.json")
firebase_admin.initialize_app(cred)

# Spotify API Credentials
CLIENT_ID = "7b31e43463f645a792cfa502547eb87a"
CLIENT_SECRET = "26fa4934017f448d86ca2a7c502e0e1c"

sp = spotipy.Spotify(auth_manager=SpotifyClientCredentials(client_id=CLIENT_ID, client_secret=CLIENT_SECRET))

def get_youtube_api_key():
    """Fetches the YouTube API Key from Firebase Firestore."""
    db = firestore.client()
    doc = db.collection("config").document("youtube_api").get()
    if doc.exists:
        return doc.to_dict().get("api_key")
    return None

def detect_language(album_name, track_names):
    """Detects album language based on album name and song titles."""
    hindi_keywords = ["hindi", "bollywood"]
    telugu_keywords = ["telugu", "tollywood"]
    english_keywords = ["english"]

    lower_album = album_name.lower()
    if any(lang in lower_album for lang in hindi_keywords):
        return "Hindi"
    if any(lang in lower_album for lang in telugu_keywords):
        return "Telugu"
    if any(lang in lower_album for lang in english_keywords):
        return "English"

    lower_tracks = " ".join([name.lower() for name in track_names])
    if any(lang in lower_tracks for lang in hindi_keywords):
        return "Hindi"
    if any(lang in lower_tracks for lang in telugu_keywords):
        return "Telugu"
    if any(lang in lower_tracks for lang in english_keywords):
        return "English"

    return "Telugu"  # Default fallback

def convert_to_ist(utc_time):
    """Converts UTC timestamp to Indian Standard Time (IST)."""
    if utc_time:
        try:
            utc_dt = datetime.strptime(utc_time, "%Y-%m-%dT%H:%M:%SZ").replace(tzinfo=timezone.utc)  # Convert to UTC
            ist_dt = utc_dt + timedelta(hours=5, minutes=30)  # Convert to IST (+5:30)
            return ist_dt.strftime("%Y-%m-%dT%H:%M:%SZ")  # Format as requested
        except Exception as e:
            return "Invalid Date"
    return "Unknown Date"

def fetch_youtube_videos(query, language):
    """Fetches YouTube video details for a song."""
    api_key = get_youtube_api_key()
    print(api_key)
    if not api_key:
        return {
            "title": "YouTube API Key Not Found",
            "url": "#",
            "video_id": "N/A",
            "channel_title": "N/A",
            "description": "N/A",
            "published_at": "N/A",
            "thumbnail_url": "N/A",
        }

    search_query = f"{query} {language} Movie Song"
    url = f"https://www.googleapis.com/youtube/v3/search?part=snippet&q={search_query}&type=video&maxResults=1&key={api_key}"
    
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        if "items" in data and data["items"]:
            item = data["items"][0]

            video_id = item.get("id", {}).get("videoId", "N/A")
            title = item.get("snippet", {}).get("title", "No title found")
            channel_title = item.get("snippet", {}).get("channelTitle", "Unknown")
            description = item.get("snippet", {}).get("description", "No description available")
            published_at = convert_to_ist(item.get("snippet", {}).get("publishedAt", "Unknown date"))
            thumbnail_url = item.get("snippet", {}).get("thumbnails", {}).get("high", {}).get("url", "")

            return {
                "title": title,
                "url": f"https://www.youtube.com/watch?v={video_id}",
                "video_id": video_id,
                "channel_title": channel_title,
                "description": description,
                "published_at": published_at,
                "thumbnail_url": thumbnail_url,
            }
    
    return {
        "title": "No video found",
        "url": "#",
        "video_id": "N/A",
        "channel_title": "N/A",
        "description": "N/A",
        "published_at": "N/A",
        "thumbnail_url": "N/A",
    }

def get_album_songs(album_name):
    """Retrieves album details from Spotify."""
    results = sp.search(q=album_name, type="album", limit=1)
    
    if results["albums"]["items"]:
        album = results["albums"]["items"][0]
        album_name = album["name"]
        album_url = album["external_urls"]["spotify"]
        album_id = album["id"]

        # Fetch tracks
        tracks = sp.album_tracks(album_id)
        track_names = [track["name"] for track in tracks["items"]]

        # Detect language
        album_language = detect_language(album_name, track_names)

        return album_name, album_url, album_language, tracks["items"]
    
    return None, None, None, None

def process_song(song, album_language):
    """Wrapper function for YouTube search (used in multithreading)."""
    song_title = song["name"]
    return song_title, fetch_youtube_videos(song_title, album_language)

def main():
    movie_name = "RRR"
    
    album_name, album_url, album_language, tracks = get_album_songs(movie_name)
    
    if not album_name:
        print("No album found for the given name.")
        return

    print(f"🎵 Album: {album_name} - {album_url} ({album_language})\n")
    print("🎶 Songs & YouTube Links:\n")

    # Multithreading for faster YouTube searches
    with ThreadPoolExecutor(max_workers=5) as executor:
        future_to_song = {executor.submit(process_song, track, album_language): track for track in tracks}

        for future in as_completed(future_to_song):
            song_title, youtube_result = future.result()

            print(f"🎵 {song_title} ({album_language})")
            print(f"   🎬 YouTube: {youtube_result['title']} - {youtube_result['url']}")
            print(f"   📺 Channel: {youtube_result['channel_title']}")
            print(f"   📝 Description: {youtube_result['description'][:100]}...")  # Truncated for readability
            print(f"   📅 Published At (IST): {youtube_result['published_at']}")
            print(f"   🖼️ Thumbnail: {youtube_result['thumbnail_url']}\n")

if __name__ == "__main__":
    main()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time
import json
 
# Setup Selenium with Brave Browser
options = Options()
options.binary_location = r"C:\Users\mahes\AppData\Local\BraveSoftware\Brave-Browser\Application\brave.exe"
options.add_argument("--headless")  # Run without opening browser
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
 
# Initialize WebDriver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)
 
# YouTube Music Channel URL
channel_url = "https://www.youtube.com/channel/UC-9-kyTW8ZkZNDHQJ6FgpwQ"
 
# Open Channel Page
driver.get(channel_url)
 
# Wait for page to load completely
WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
time.sleep(3)
 
# Parse the page with BeautifulSoup
soup = BeautifulSoup(driver.page_source, "html.parser")
 
# Dictionary to store headings and their links
data = {}
 
# Find all sections with headings and their associated links
sections = soup.find_all("h2")
for section in sections:
    heading = section.text.strip()
    links = []
   
    # Find the next sibling elements that contain links
    next_element = section.find_next()
    while next_element and next_element.name != "h2":  # Stop at the next heading
        for a_tag in next_element.find_all("a", href=True):
            href = a_tag["href"]
            if href.startswith("/watch") or href.startswith("/playlist") or href.startswith("/channel"):
                full_link = "https://www.youtube.com" + href
                if full_link not in links:  # Avoid duplicates
                    links.append(full_link)
        next_element = next_element.find_next()
   
    # Add to dictionary
    data[heading] = links if links else ["No links found under this heading"]
 
# Convert to JSON and print
json_output = json.dumps(data, indent=2, ensure_ascii=False)
print(json_output)
 
# Optionally, save to a file
with open("youtube_channel_data.json", "w", encoding="utf-8") as f:
    f.write(json_output)
 
# Close the browser
driver.quit()
 
```

```python
import time
import json
from datetime import datetime
from urllib.parse import urlparse, parse_qs

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from ytmusicapi import YTMusic

ytmusic = YTMusic()


def format_youtube_release_date(iso_date_str):
    try:
        dt = datetime.fromisoformat(iso_date_str)
        return dt.strftime("%d-%m-%Y %H:%M:%S")
    except Exception:
        return "N/A"


def setup_driver():
    options = Options()
    options.binary_location = r"C:\Users\mahes\AppData\Local\BraveSoftware\Brave-Browser\Application\brave.exe"
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920x1080")
    options.add_argument("--disable-extensions")
    options.add_argument("--log-level=3")
    return webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)


def scrape_latest_telugu_songs():
    search_query = "latest Telugu songs"
    url = f"https://www.youtube.com/results?search_query={search_query.replace(' ', '+')}"

    driver = setup_driver()
    driver.get(url)

    time.sleep(2)
    for _ in range(3):
        driver.find_element(By.TAG_NAME, "body").send_keys(Keys.END)
        time.sleep(1)

    video_links = driver.find_elements(By.CSS_SELECTOR, "a#video-title")
    videos = []
    seen = set()

    for video in video_links:
        title = video.get_attribute("title")
        href = video.get_attribute("href")

        if not href or "/watch" not in href:
            continue

        parsed = urlparse(href)
        video_id = parse_qs(parsed.query).get("v", [None])[0]
        if not video_id:
            continue

        try:
            song_info = ytmusic.get_song(video_id)
            song_title = song_info["videoDetails"]["title"]
            song_artist = song_info["videoDetails"]["author"]

            # Skip duplicates
            if (song_title, song_artist) in seen:
                continue
            seen.add((song_title, song_artist))

            duration = song_info["videoDetails"]["lengthSeconds"]
            views = song_info["videoDetails"]["viewCount"]
            thumbnail = song_info["videoDetails"]["thumbnail"]["thumbnails"][-1]["url"]
            release_date = (
                song_info.get("microformat", {})
                .get("microformatDataRenderer", {})
                .get("publishDate", None)
            )
            release_date = format_youtube_release_date(release_date) if release_date else "N/A"

            videos.append({
                "title": song_title,
                "video_url": href,
                "artist": song_artist,
                "duration": duration,
                "views": views,
                "thumbnail_url": thumbnail,
                "release_date": release_date
            })

            if len(videos) >= 20:
                break

        except Exception as e:
            print(f"Error processing video ID {video_id}: {e}")
            continue

    driver.quit()

    result_json = {
        "query": search_query,
        "results": videos
    }

    with open("latest_telugu_songs.json", "w", encoding="utf-8") as f:
        json.dump(result_json, f, indent=2, ensure_ascii=False)

    return result_json


if __name__ == "__main__":
    data = scrape_latest_telugu_songs()
    print(json.dumps(data, indent=2, ensure_ascii=False))
```

```python
import time
import json
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

SPOTIFY_PLAYLIST_URLS = [
    "https://open.spotify.com/playlist/37i9dQZF1DX5VOFoIqmrOV",
    "https://open.spotify.com/playlist/37i9dQZF1DX0TyiNWW7uUQ",
    "https://open.spotify.com/playlist/37i9dQZF1DWXtlo6ENS92N",
    "https://open.spotify.com/playlist/37i9dQZF1DX5cZuAHLNjGz",
    "https://open.spotify.com/playlist/37i9dQZF1DX5q67ZpWyRrZ"
]

def setup_driver():
    options = Options()
    options.add_argument("--headless")  # Run in headless mode
    options.add_argument("--disable-gpu")
    options.add_argument("--window-size=1920x1080")
    options.add_argument("--disable-extensions")
    options.add_argument("--log-level=3")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")

    return webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

def scrape_spotify_playlist(url):
    driver = setup_driver()
    driver.get(url)

    print(f"[🔎] Scraping: {url}")
    wait = WebDriverWait(driver, 15)

    try:
        # Title is under data-testid="entityTitle"
        playlist_title_el = wait.until(EC.presence_of_element_located(
            (By.CSS_SELECTOR, 'span[data-testid="entityTitle"]')
        ))
        playlist_title = playlist_title_el.text
    except Exception as e:
        print(f"[❌] Playlist title not found: {e}")
        playlist_title = "Unknown Playlist"

    # Scroll to load more songs
    scroll_pause_time = 1.5
    for _ in range(5):
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(scroll_pause_time)

    tracks = []
    rows = driver.find_elements(By.CSS_SELECTOR, 'div[data-testid="tracklist-row"]')
    print(f"[🎵] Found {len(rows)} tracks")

    for row in rows:
        try:
            title = row.find_element(By.CSS_SELECTOR, 'div[aria-colindex="2"] span').text
            artist = row.find_element(By.CSS_SELECTOR, 'div[aria-colindex="2"] a').text
            duration = row.find_element(By.CSS_SELECTOR, 'div[aria-colindex="5"]').text
            album_art = row.find_element(By.CSS_SELECTOR, 'img').get_attribute("src")

            track_info = {
                "title": title,
                "artist": artist,
                "duration": duration,
                "album_art": album_art
            }
            tracks.append(track_info)
        except Exception as e:
            print(f"[⚠️] Skipped track due to error: {e}")

    driver.quit()

    return {
        "playlist_url": url,
        "playlist_title": playlist_title,
        "total_tracks": len(tracks),
        "tracks": tracks
    }

def main():
    result = []
    for url in SPOTIFY_PLAYLIST_URLS:
        data = scrape_spotify_playlist(url)
        result.append(data)

    with open("spotify_playlists.json", "w", encoding="utf-8") as f:
        json.dump(result, f, indent=2, ensure_ascii=False)

    print("[✅] All playlists scraped and saved to spotify_playlists.json")

if __name__ == "__main__":
    main()
```

```python
import time
import json
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

SPOTIFY_PLAYLIST_URLS = [
    "https://open.spotify.com/playlist/37i9dQZF1DX5VOFoIqmrOV",
    "https://open.spotify.com/playlist/37i9dQZF1DX0TyiNWW7uUQ",
    "https://open.spotify.com/playlist/37i9dQZF1DWXtlo6ENS92N",
    "https://open.spotify.com/playlist/37i9dQZF1DX5cZuAHLNjGz",
    "https://open.spotify.com/playlist/37i9dQZF1DX5q67ZpWyRrZ"
]

def setup_driver():
    options = Options()
    options.add_argument("--start-maximized")
    # options.add_argument("--headless")  # Enable this if you want headless mode
    return webdriver.Chrome(options=options)

def scroll_until_all_tracks_loaded(driver):
    SCROLL_PAUSE_TIME = 2
    last_height = driver.execute_script("return document.body.scrollHeight")

    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(SCROLL_PAUSE_TIME)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height

    tracks = driver.find_elements(By.CSS_SELECTOR, 'div[data-testid="tracklist-row"]')
    return len(tracks)

def scrape_spotify_playlist(url):
    driver = setup_driver()
    driver.get(url)
    print(f"[🔎] Scraping: {url}")

    try:
        playlist_title_elem = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "h1"))
        )
        playlist_title = playlist_title_elem.text
    except Exception as e:
        print(f"[❌] Failed to get playlist title: {e}")
        driver.quit()
        return {
            "playlist_title": "Unknown",
            "playlist_url": url,
            "total_tracks": 0,
            "tracks": []
        }

    scroll_until_all_tracks_loaded(driver)
    track_rows = driver.find_elements(By.CSS_SELECTOR, 'div[data-testid="tracklist-row"]')
    print(f"[🎵] Found {len(track_rows)} tracks")

    tracks_data = []

    for track in track_rows:
        try:
            name_elem = track.find_element(By.CSS_SELECTOR, 'div[dir="auto"]:nth-child(2)')
            artist_elem = track.find_elements(By.CSS_SELECTOR, 'span[class*="Type__StyledComponent"]')[1]

            name = name_elem.text
            artist = artist_elem.text
            tracks_data.append({
                "title": name,
                "artist": artist
            })
        except Exception as e:
            print(f"[⚠️] Skipped track due to error: {e}")
            continue

    driver.quit()

    return {
        "playlist_title": playlist_title,
        "playlist_url": url,
        "total_tracks": len(tracks_data),
        "tracks": tracks_data
    }

def main():
    result = []
    for url in SPOTIFY_PLAYLIST_URLS:
        data = scrape_spotify_playlist(url)
        result.append(data)

    with open("spotify_playlists.json", "w", encoding="utf-8") as f:
        json.dump(result, f, ensure_ascii=False, indent=2)

    print("[✅] All playlists scraped and saved to spotify_playlists.json")

if __name__ == "__main__":
    main()
```
