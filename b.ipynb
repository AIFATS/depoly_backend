```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time
import json

# Setup Selenium with Brave Browser
options = Options()
options.binary_location = r"C:\Users\ABHISHEK\AppData\Local\BraveSoftware\Brave-Browser\Application\brave.exe"
options.add_argument("--headless")  # Run in headless mode
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

# Initialize WebDriver
print("Initializing WebDriver...")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# YouTube Music Channel URL
channel_url = "https://www.youtube.com/channel/UC-9-kyTW8ZkZNDHQJ6FgpwQ"

# Open Channel Page
print(f"Opening YouTube Channel: {channel_url}")
driver.get(channel_url)

# Scroll down to load all elements
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(3)  # Wait for additional content to load

# Wait for thumbnails to be present
print("Waiting for thumbnails to load...")
WebDriverWait(driver, 10).until(
    EC.presence_of_all_elements_located((By.TAG_NAME, "img"))
)

# Parse the page with BeautifulSoup
print("Parsing the HTML content using BeautifulSoup...")
soup = BeautifulSoup(driver.page_source, "html.parser")

# Dictionary to store headings, playlist names, links, and thumbnails
data = {}

# Find all section headings (h2 tags)
sections = soup.find_all("h2")
print(f"\nFound {len(sections)} section headings.\n")

for section in sections:
    heading = section.text.strip()  # Extract heading text
    playlists = []  # List for storing playlist details

    # Find the next elements that contain playlist links
    next_element = section.find_next()
    while next_element and next_element.name != "h2":  # Stop at the next heading
        for a_tag in next_element.find_all("a", href=True):
            href = a_tag["href"]

            # Ensure it's a playlist link
            if href.startswith("/playlist"):
                full_link = "https://www.youtube.com" + href

                # Find the correct playlist name
                playlist_name_tag = a_tag.find_next("span")  # Get the correct playlist name
                if playlist_name_tag:
                    playlist_name = playlist_name_tag.text.strip()

                    # Move **down** to find the thumbnail
                    thumbnail_url = "No Thumbnail"
                    song_count_element = a_tag.find_next("yt-thumbnail-overlay-badge-view-model")
                    if song_count_element:
                        img_tag = song_count_element.find_next("img")  # Find the image below count
                        if img_tag and img_tag.has_attr("src"):
                            thumbnail_url = img_tag["src"]

                    # Ensure it's a valid name and avoid duplicates
                    if playlist_name and not any(p["link"] == full_link for p in playlists):
                        playlists.append({
                            "name": playlist_name,
                            "link": full_link,
                            "thumbnail": thumbnail_url
                        })

        next_element = next_element.find_next()

    # Store data in dictionary
    data[heading] = playlists if playlists else [{"message": "No playlists found under this heading"}]

    # Print extracted details
    print("\n====================")
    print(heading)  # Print main heading
    print("====================")
    if playlists:
        for playlist in playlists:
            print(f"  🎵 {playlist['name']}: {playlist['link']}")
            print(f"  🖼 Thumbnail: {playlist['thumbnail']}\n")
    else:
        print("  ➜ No playlists found under this heading.")
    print()  # Blank line for spacing

# Convert the dictionary to JSON format
json_output = json.dumps(data, indent=2, ensure_ascii=False)

# Save the extracted data to a JSON file
with open("youtube_channel_data.json", "w", encoding="utf-8") as f:
    f.write(json_output)

print("\nData has been saved to 'youtube_channel_data.json'.")

# Close the browser
driver.quit()
print("Web scraping complete. Browser closed.")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time
import json

# Setup Selenium with Brave Browser
options = Options()
options.binary_location = r"C:\Users\ABHISHEK\AppData\Local\BraveSoftware\Brave-Browser\Application\brave.exe"
options.add_argument("--headless")  # Run in headless mode
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

# Initialize WebDriver
print("Initializing WebDriver...")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# YouTube Music Channel URL
channel_url = "https://www.youtube.com/channel/UC-9-kyTW8ZkZNDHQJ6FgpwQ"

# Open Channel Page
print(f"Opening YouTube Channel: {channel_url}")
driver.get(channel_url)

# Wait for the page to load completely
print("Waiting for page to fully load...")
WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
time.sleep(3)  # Allow extra time for dynamic content to load

# Parse the page with BeautifulSoup
print("Parsing the HTML content using BeautifulSoup...")
soup = BeautifulSoup(driver.page_source, "html.parser")

# Dictionary to store headings, playlists, and videos
data = {}

# Find all section headings (h2 tags)
sections = soup.find_all("h2")
print(f"\nFound {len(sections)} section headings.\n")
playlist_count = 0

for section in sections:
    if playlist_count >= 10:
        break
    
    heading = section.text.strip()  # Extract heading text
    playlists = {}  # Dictionary for playlist names and their data

    # Find the next elements that contain playlist links
    next_element = section.find_next()
    while next_element and next_element.name != "h2":  # Stop at the next heading
        for a_tag in next_element.find_all("a", href=True):
            if playlist_count >= 10:
                break
            
            href = a_tag["href"]
            
            # Ensure it's a playlist link
            if href.startswith("/playlist"):
                full_link = "https://www.youtube.com" + href
                
                # Find the correct playlist name
                playlist_name_tag = a_tag.find_next("span")
                if playlist_name_tag:
                    playlist_name = playlist_name_tag.text.strip()
                    
                    # Ensure it's a valid name and avoid duplicates
                    if playlist_name and playlist_name not in playlists:
                        print(f"Fetching videos for playlist: {playlist_name}")
                        
                        # Open the playlist page
                        driver.get(full_link)
                        time.sleep(3)  # Wait for the page to load
                        playlist_soup = BeautifulSoup(driver.page_source, "html.parser")
                        
                        # Find video links and titles (limit to 5 videos)
                        videos = []
                        video_elements = playlist_soup.find_all("a", href=True)
                        seen_videos = set()
                        count = 0
                        
                        for video in video_elements:
                            if count >= 5:
                                break
                            video_href = video["href"]
                            if video_href.startswith("/watch"):
                                video_url = "https://www.youtube.com" + video_href
                                
                                # Ensure unique videos
                                if video_url in seen_videos:
                                    continue
                                seen_videos.add(video_url)
                                
                                # Extract title (tag1: yt-formatted-string, tag2: span)
                                video_title_tag1 = video.find("yt-formatted-string", class_="style-scope ytd-watch-metadata")
                                video_title = video_title_tag1["title"].strip() if video_title_tag1 and video_title_tag1.has_attr("title") else None

                                if not video_title:
                                    video_title_tag2 = video.find("span", {"id": "video-title"})
                                    video_title = video_title_tag2["title"].strip() if video_title_tag2 and video_title_tag2.has_attr("title") else "Unknown Title"
                                
                                videos.append({
                                    "title": video_title,
                                    "video_url": video_url
                                })
                                count += 1
                        
                        playlists[playlist_name] = {
                            "playlist_url": full_link,
                            "videos": videos
                        }
                        playlist_count += 1
        
        next_element = next_element.find_next()
    
    # Store data in dictionary
    if playlists:
        data[heading] = playlists

# Convert the dictionary to JSON format
json_output = json.dumps(data, indent=2, ensure_ascii=False)

# Save the extracted data to a JSON file
with open("youtube_channel_data.json", "w", encoding="utf-8") as f:
    f.write(json_output)

print("\nData has been saved to 'youtube_channel_data.json'.")

# Close the browser
driver.quit()
print("Web scraping complete. Browser closed.")
```
